## Task 1: Encapsulation for Student class
All three attributes (`__studentID`, `__age`, `__name`) are private. They can only be read or changed through the getter and setter functions.

```python
class Student:
    def __init__(self, studentID, name, age):
        self.__studentID = studentID
        self.__name = name
        self.__age = age

    def getID(self):
        return self.__studentID

    def setID(self, studentID):
        self.__studentID = studentID

    def getName(self):
        return self.__name

    def setName(self, name):
        self.__name = name

    def getAge(self):
        return self.__age

    def setAge(self, age):
        if age > 0:                  # simple check so age cannot be negative
            self.__age = age
        else:
            print('Invalid age!')
```

```python
s = Student('P1234567', 'John Tan', 18)
print(s.getID(), s.getName(), s.getAge())

s.setName('Mary Lim')
s.setAge(19)
s.setAge(-5)                         # rejected by the setter
print(s.getID(), s.getName(), s.getAge())

# The private variables cannot be reached directly from outside
try:
    print(s.__name)
except AttributeError as e:
    print('Error:', e)
```

## Task 2: Deriving a Mouse class from abstract class Animal
`Animal` is abstract (it uses `ABC` and an `@abstractmethod`), so it cannot be created on its own. `Cat`, `Dog` and `Mouse` each give their own version of `makeSound()`.

```python
from abc import ABC, abstractmethod

class Animal(ABC):
    def __init__(self, name):
        self._name = name

    def getName(self):
        return self._name

    @abstractmethod
    def makeSound(self):
        pass

    def __str__(self):
        return f'{self.__class__.__name__} {self._name} says {self.makeSound()}'


class Cat(Animal):
    def makeSound(self):
        return 'Meow'


class Dog(Animal):
    def makeSound(self):
        return 'Woof'


class Mouse(Animal):
    def makeSound(self):
        return 'Squeak'
```

```python
# Main program to test the Mouse class
animals = [Cat('Tom'), Dog('Spike'), Mouse('Jerry')]
for animal in animals:
    print(animal)

m = Mouse('Mickey')
print(m.getName(), 'is a Mouse?', isinstance(m, Mouse))
print(m.getName(), 'is an Animal?', isinstance(m, Animal))

# Animal is abstract, so this should fail
try:
    a = Animal('Generic')
except TypeError as e:
    print('Error:', e)
```

## Original Stack class (Appendix A)
Needed for Task 4, which inherits from it.

```python
class Stack:
    def __init__(self):
        self.__list = []

    def isEmpty(self):
        return self.__list == []

    def size(self):
        return len(self.__list)

    def clear(self):
        self.__list.clear()

    def push(self, item):
        self.__list.append(item)

    def pop(self):
        if self.isEmpty():
            return None
        else:
            return self.__list.pop()

    def get(self):
        if self.isEmpty():
            return None
        else:
            return self.__list[-1]

    def __str__(self):
        output = '<'
        for i in range(len(self.__list)):
            item = self.__list[i]
            if i < len(self.__list) - 1:
                output += f'{str(item)}, '
            else:
                output += f'{str(item)}'
        output += '>'
        return output
```

## Task 3: Stack2 (written from scratch)
The top of the stack is now the **left** side (index 0). `push` inserts at index 0 and `pop` removes index 0, so printing the list shows the top first.

```python
class Stack2:
    def __init__(self):
        self.__list = []

    def isEmpty(self):
        return self.__list == []

    def size(self):
        return len(self.__list)

    def clear(self):
        self.__list.clear()

    def push(self, item):
        self.__list.insert(0, item)        # top is at the left

    def pop(self):
        if self.isEmpty():
            return None
        else:
            return self.__list.pop(0)

    def get(self):
        if self.isEmpty():
            return None
        else:
            return self.__list[0]

    def __str__(self):
        output = '<'
        for i in range(len(self.__list)):
            item = self.__list[i]
            if i < len(self.__list) - 1:
                output += f'{str(item)}, '
            else:
                output += f'{str(item)}'
        output += '>'
        return output
```

```python
s = Stack2()
print(s.pop())
for i in range(1,6):
    s.push(i)
print('Content of stack = ',s)
print('Item at top = ',s.get())
print('Size = ', s.size())
while not s.isEmpty():
    print(s.pop())
    print(s)
```

## Task 4: Stack3 (inherits from Stack)
`push`, `pop`, `get`, `isEmpty`, `size` and `clear` are all reused from `Stack`. The data is stored the same way (top at the right of the list); only `__str__` is overridden so it prints the items in reverse, with the top on the left.

Since `Stack.__list` is private (name mangled to `_Stack__list`), Stack3 builds the output by popping everything into a temporary list and pushing it back, using only the public methods of `Stack`.

```python
class Stack3(Stack):
    def __str__(self):
        # Take the items off the top, one by one (top comes out first)
        items = []
        while not self.isEmpty():
            items.append(self.pop())

        # Put them back so the stack is unchanged
        for i in range(len(items) - 1, -1, -1):
            self.push(items[i])

        output = '<'
        for i in range(len(items)):
            if i < len(items) - 1:
                output += f'{str(items[i])}, '
            else:
                output += f'{str(items[i])}'
        output += '>'
        return output
```

```python
s = Stack3()
print(s.pop())
for i in range(1,6):
    s.push(i)
print('Content of stack = ',s)
print('Item at top = ',s.get())
print('Size = ', s.size())
while not s.isEmpty():
    print(s.pop())
    print(s)
```

**Which method is better?**

* **From scratch (Stack2):** full control and simple to read, but every method has to be written (and tested) again even though most of them are the same as in `Stack`.
* **Inheritance (Stack3):** only the one method that is different (`__str__`) is written; everything else is reused, so there is less code and less chance of bugs. Fixes made in `Stack` are picked up by `Stack3` automatically. Also, `push`/`pop` stay at the end of the Python list, which is O(1), while Stack2 inserts / removes at index 0, which is O(n).

I prefer **inheritance** when the new class really "is a" kind of the old class and only a small part of its behaviour changes.

## Task 5: Queue class
Items join at the back (right) and leave from the front (left), so it is first in first out (FIFO).

```python
class Queue:
    def __init__(self):
        self.__list = []

    def isEmpty(self):
        return self.__list == []

    def size(self):
        return len(self.__list)

    def clear(self):
        self.__list.clear()

    def enqueue(self, item):
        self.__list.append(item)           # join at the back

    def dequeue(self):
        if self.isEmpty():
            return None
        else:
            return self.__list.pop(0)      # leave from the front

    def get(self):
        if self.isEmpty():
            return None
        else:
            return self.__list[0]

    def __str__(self):
        output = '<'
        for i in range(len(self.__list)):
            item = self.__list[i]
            if i < len(self.__list) - 1:
                output += f'{str(item)}, '
            else:
                output += f'{str(item)}'
        output += '>'
        return output
```

```python
q = Queue()
print(q.dequeue())
for i in range(1,6):
    q.enqueue(i)
print('Content of queue is = ',q)
print('Item at front = ',q.get())
print('Size = ', q.size())
while not q.isEmpty():
    print(q.dequeue())
    print(q)
```

## Task 6: Deque class
A double ended queue: items can be added or removed at both the head (left) and the tail (right).

```python
class Deque:
    def __init__(self):
        self.__list = []

    def isEmpty(self):
        return self.__list == []

    def size(self):
        return len(self.__list)

    def clear(self):
        self.__list.clear()

    def addHead(self, item):
        self.__list.insert(0, item)

    def addTail(self, item):
        self.__list.append(item)

    def removeHead(self):
        if self.isEmpty():
            return None
        else:
            return self.__list.pop(0)

    def removeTail(self):
        if self.isEmpty():
            return None
        else:
            return self.__list.pop()

    def getHead(self):
        if self.isEmpty():
            return None
        else:
            return self.__list[0]

    def getTail(self):
        if self.isEmpty():
            return None
        else:
            return self.__list[-1]

    def __str__(self):
        output = '<'
        for i in range(len(self.__list)):
            item = self.__list[i]
            if i < len(self.__list) - 1:
                output += f'{str(item)}, '
            else:
                output += f'{str(item)}'
        output += '>'
        return output
```

```python
d = Deque()
print(d.removeHead())
print(d.removeTail())
for i in range(1,6):
    d.addTail(i)
print('Content of deque is = ',d)
print('Item at front = ',d.getHead())
print('Item at back = ',d.getTail())
print('Size = ', d.size())
count=0
while not d.isEmpty():
    # Alternating remove head and tail
    if count % 2 ==0:
        print(d.removeHead())
    else:
        print(d.removeTail())
    print(d)
    count+=1
```

## Node and SortedList (Appendix B)
One small fix: in the original, `insert` adds 1 to `length` and then `appendToHead` adds 1 again, so a new head was counted twice. Here `insert` only counts in the cases that don't call `appendToHead`.

```python
class Node:
    # Constructor
    def __init__(self):
        self.nextNode = None


class SortedList:
    def __init__(self):
        self.headNode = None
        self.currentNode = None
        self.length = 0

    def __appendToHead(self, newNode):
        oldHeadNode = self.headNode
        self.headNode = newNode
        self.headNode.nextNode = oldHeadNode
        self.length += 1

    def insert(self, newNode):
        # If list is currently empty
        if self.headNode == None:
            self.headNode = newNode
            self.length += 1
            return

        # Check if it is going to be new head
        if newNode < self.headNode:
            self.__appendToHead(newNode)       # this adds 1 to length
            return

        self.length += 1
        # Check it is going to be inserted
        # between any pair of Nodes (left, right)
        leftNode = self.headNode
        rightNode = self.headNode.nextNode
        while rightNode != None:
            if newNode < rightNode:
                leftNode.nextNode = newNode
                newNode.nextNode = rightNode
                return
            leftNode = rightNode
            rightNode = rightNode.nextNode

        # Once we reach here it must be added at the tail
        leftNode.nextNode = newNode

    def __str__(self):
        # We start at the head
        output = ""
        node = self.headNode
        firstNode = True
        while node != None:
            if firstNode:
                output = node.__str__()
                firstNode = False
            else:
                output += (',' + node.__str__())
            node = node.nextNode
        return output
```

## Task 7: Point class in a SortedList
Sort by `x` first; if the `x` values are the same, sort by `y`.

```python
class Point(Node):
    def __init__(self, x, y):
        self.x = x
        self.y = y
        super().__init__()

    def __eq__(self, otherNode):
        if otherNode == None:
            return False
        else:
            return self.x == otherNode.x and self.y == otherNode.y

    def __lt__(self, otherNode):
        if otherNode == None:
            raise TypeError("'<' not supported between instances of 'Point' and 'NoneType'")
        if self.x == otherNode.x:
            return self.y < otherNode.y        # same x, so compare y
        return self.x < otherNode.x

    def __str__(self):
        return f'({self.x},{self.y})'
```

```python
import random
l = SortedList()
maxNum = 10
for i in range(0,maxNum):
    point = Point(random.randint(1,maxNum),
                  random.randint(1,maxNum) )
    l.insert(point)
print(l)
```

## Task 8(a): Fruit class in a SortedList
Sort by length of the name first; if the lengths are the same, sort alphabetically (ignoring upper / lower case).

To print the list in the same format as the expected output (`['Date', 'lime', ...]`), `Fruit.__str__` puts quotes around the name and `SortedList.__str__` is updated to join the items with `', '` inside square brackets.

```python
class Fruit(Node):
    def __init__(self, name):
        self.name = name
        super().__init__()

    def __eq__(self, otherNode):
        if otherNode == None:
            return False
        else:
            return self.name == otherNode.name

    def __lt__(self, otherNode):
        if otherNode == None:
            raise TypeError("'<' not supported between instances of 'Fruit' and 'NoneType'")
        if len(self.name) == len(otherNode.name):
            return self.name.lower() < otherNode.name.lower()   # same length, so alphabetical
        return len(self.name) < len(otherNode.name)

    def __str__(self):
        return f"'{self.name}'"


class SortedList:
    def __init__(self):
        self.headNode = None
        self.currentNode = None
        self.length = 0

    def __appendToHead(self, newNode):
        oldHeadNode = self.headNode
        self.headNode = newNode
        self.headNode.nextNode = oldHeadNode
        self.length += 1

    def insert(self, newNode):
        # If list is currently empty
        if self.headNode == None:
            self.headNode = newNode
            self.length += 1
            return

        # Check if it is going to be new head
        if newNode < self.headNode:
            self.__appendToHead(newNode)
            return

        self.length += 1
        # Check it is going to be inserted
        # between any pair of Nodes (left, right)
        leftNode = self.headNode
        rightNode = self.headNode.nextNode
        while rightNode != None:
            if newNode < rightNode:
                leftNode.nextNode = newNode
                newNode.nextNode = rightNode
                return
            leftNode = rightNode
            rightNode = rightNode.nextNode

        # Once we reach here it must be added at the tail
        leftNode.nextNode = newNode

    def __str__(self):
        # Show the list like a Python list: [item1, item2, ...]
        output = "["
        node = self.headNode
        firstNode = True
        while node != None:
            if firstNode:
                output += node.__str__()
                firstNode = False
            else:
                output += (', ' + node.__str__())
            node = node.nextNode
        output += "]"
        return output
```

```python
l = SortedList()
# Populate a list with fruit names
fruits = ['Cherry', 'Apricot', 'lime', 'blueberry', 'Apple', 'Date']
print('Before sorting')
print(fruits)
for fruit in fruits:
    l.insert(Fruit(fruit))
print('\nAfter sorting')
print(l)
```

## Task 8(b): Reading fruits from a file
Two new functions are added to `SortedList` so we can walk through it one node at a time:
* `resetForIteration()` moves `currentNode` back to the head and returns it.
* `nextNode()` moves `currentNode` one step forward and returns it (`None` at the end).

`SortedList2` inherits everything else from the `SortedList` above.

```python
class SortedList2(SortedList):
    def resetForIteration(self):
        self.currentNode = self.headNode
        return self.currentNode

    def nextNode(self):
        if self.currentNode != None:
            self.currentNode = self.currentNode.nextNode
        return self.currentNode
```

```python
l = SortedList2()
# Read fruits from a file and sort in list
f = open('fruits.txt', 'r')
for line in f:
    fruit = line.strip()
    l.insert(Fruit( fruit ))
f.close()

# Write sorted fruits into file
f = open('fruits_sorted.txt', 'w')
fruit = l.resetForIteration()
while fruit != None:
    f.write(fruit.name+"\n")
    fruit = l.nextNode()
f.close()
```
